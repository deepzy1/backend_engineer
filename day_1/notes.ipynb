Shallow Copy vs Deep Copy

Shallow copy = New outer container, shared nested objects.

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item(10))
print(add_item(20))
print(add_item(30))
```

```python
def add_item(item, items=None):
    # print(items)
    if items is None:
        items = []

    items.append(item)
    return items

print(add_item(10,[1,2,3]))
print(add_item(20))
print(add_item(30))
```
