```python
#madlibs game

import random

stories = [
    {
        "template": (
            "Yesterday, I met a {adjective} {animal} on {place}. "
            "Suddenly, it started to {verb} in front of {number} "
            "confused people. Everyone screamed, 'What is happening?!'"
        ),
        "questions": {
            "adjective": "Give me an adjective: ",
            "animal": "Name an animal: ",
            "place": "Name a place: ",
            "verb": "Enter a verb: ",
            "number": "Enter a number: "
        }
    },
    {
        "template": (
            "A {adjective} {animal} became the president of {place}. "
            "Its first rule was to make everyone {verb} "
            "{number} times every morning!"
        ),
        "questions": {
            "adjective": "Give me an adjective: ",
            "animal": "Name an animal: ",
            "place": "Name a place: ",
            "verb": "Enter a verb: ",
            "number": "Enter a number: "
        }
    }
]

story = random.choice(stories)
answers = {}

print("\n🎮 Welcome to Mad Libs!\n")

for word, question in story["questions"].items():
    answers[word] = input(question).strip()

print("\n📖 YOUR CRAZY STORY\n")
print(story["template"].format(**answers))

```
