```python
#2d list

fruits=["apple","banana","cocunut","berries"]
veggies=["carrot","beetroot","cabbage"]
meat=["chicken","mutton","fis","turkey"]

grociers=[fruits,veggies,meat]

for collection in grociers:
    for food in collection:
        print(food,end='  ')
    print()
```

```python
#nampad

numpad=((1,2,3),
        (4,5,6),
        (7,8,9),
        ("*",0,"#"))

for row in numpad:
    for num in row:
        print(num,end="|")
    print("")
```
